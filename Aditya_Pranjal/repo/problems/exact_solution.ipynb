```python
import math
import torch

def exact_heat_ic(x, k_max=16):
    """Initial condition for the heat equation[cite: 2]."""
    u0 = torch.zeros_like(x)
    torch.manual_seed(42) 
    for k in range(1, k_max + 1):
        a_k = torch.randn(1) * (k**-2)
        b_k = torch.randn(1) * (k**-2)
        u0 += a_k * torch.sin(2 * math.pi * k * x) + b_k * torch.cos(2 * math.pi * k * x)
    return u0

def exact_advection_ic(x, k_max=16):
    """Initial condition for the advection equation[cite: 2]."""
    return exact_heat_ic(x, k_max)

def exact_water_wave_ic(x, z, a=0.01, g=9.81, h=0.159, L=1.0):
    """Initial velocity potential for linear water waves[cite: 2]."""
    k = 2 * math.pi / L
    omega = math.sqrt(g * k * math.tanh(k * h))
    phi0 = (a * g / omega) * (torch.cosh(k * (z + h)) / math.cosh(k * h)) * torch.sin(k * x)
    return phi0

def exact_oscillator(t, w0=2.0, zeta=0.1, z0=1.0, v0=0.0):
    """Exact closed-form solution for the underdamped oscillator[cite: 5]."""
    wd = w0 * math.sqrt(1 - zeta**2)
    A = z0
    B = (v0 + zeta * w0 * z0) / wd
    
    decay = torch.exp(-zeta * w0 * t)
    oscillation = A * torch.cos(wd * t) + B * torch.sin(wd * t)
    z = decay * oscillation
    z_dot = -zeta * w0 * z + decay * (-A * wd * torch.sin(wd * t) + B * wd * torch.cos(wd * t))
    return torch.cat([z, z_dot], dim=1)
```
